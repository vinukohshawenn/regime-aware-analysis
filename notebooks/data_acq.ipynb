```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = {
    'gold':     'GC=F',
    'silver':   'SI=F',
    'vix':      '^VIX',
    'interest': '^TNX',
    'usd_idx':  'DX-Y.NYB'
}

ticker_symbols = list(tickers.values())
df = yf.download(ticker_symbols, start='2010-02-01', end='2026-06-15')['Close']

reverse_tickers = {v: k for k, v in tickers.items()}
df = df.rename(columns=reverse_tickers).dropna()

# Compute returns
df['au_returns'] = df['gold'].pct_change().apply(lambda x: np.log(1+x))
df['ag_returns'] = df['silver'].pct_change().apply(lambda x: np.log(1+x))

# Rolling stats
df['au_vol']  = df['au_returns'].rolling(21).std()
df['ag_vol']  = df['ag_returns'].rolling(21).std()
df['au_mean'] = df['au_returns'].rolling(21).mean()
df['ag_mean'] = df['ag_returns'].rolling(21).mean()
df = df.dropna()

df.index.name = 'Date'
df = df.reset_index()


df['ID_Column'] = df.index

cols = ['ID_Column', 'Date'] + [c for c in df.columns if c not in ['ID_Column', 'Date']]
df = df[cols]

df.to_csv('gold_silver_data.csv', index=False)

print(f"Saved. Gold range: {df['gold'].min():.0f}–{df['gold'].max():.0f}")
print("Check the CSV. You now have a hardcoded 'ID_Column' and a hardcoded 'Date' column.")
```
